# Fitting an XRD pattern with `iobs_ngc`

This notebook walks through a full Ruland–Smarsly refinement of a non-graphitic
carbon scattering pattern using the high-level `FitPattern` API.

We use the bundled `example_fit.xy` as the experimental data. Its first column
is already the scattering vector `s` (Å⁻¹) rather than 2θ, so we pass
`x_is_s=True` when loading it.

```python
from iobs_ngc import FitPattern
```

## Instrument-dependent parameters

`par_r` (goniometer radius, cm), `par_delta` (divergence angle, °) and
`par_l` (irradiated sample length, cm) describe the measurement geometry. They
are **not refined** during fitting and **must be set to match your instrument**.
Here we supply them inline as a dict, using the default example values. Any
keys we don't provide are filled from the bundled defaults.

```python
instrument_params = {"par_r": 24, "par_delta": 0.5, "par_l": 0.7}
```

## Load and fit

Create a `FitPattern`, passing the instrument parameters and `x_is_s=True`
(only if the data is already in `s`, if it is in 2θ this argument can be skipped), then run the multistage fit.

```python
session = FitPattern(
    "example_fit.xy",
    params=instrument_params,
    x_is_s=True,
)
session.fit()
```

```python
print(session.fitted_params)
```

## Microstructure report

`report()` returns a human-readable summary: a fit-quality header (R², RSS,
nfev), the refined parameters, and the microstructure quantities calculated
from them.

```python
print(session.report())
```

## Plot the result

Plot the experimental data, the fitted model, and the residuals.

```python
session.plot(show=True)
```

## Optional: Savitzky-Golay smoothing

Set `smooth=True` to apply a Savitzky-Golay filter to the intensity data
before fitting. This suppresses high-frequency noise in experimental patterns;
`example_fit.xy` is a clean synthetic pattern, so the visible effect here is
small. The window length and polynomial order are configurable.

```python
session_smoothed = FitPattern(
    "example_fit.xy",
    params=instrument_params,
    x_is_s=True,
    smooth=True,
    savgol_window=11,
    savgol_polyorder=3,
)
session_smoothed.fit()
session_smoothed.plot(show=True)
```
